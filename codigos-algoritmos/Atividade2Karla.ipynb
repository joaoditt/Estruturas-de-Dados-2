## PARTE 2 - Programa que contabiliza as operações realizadas durante a ordenação.

```python
import random


# Bubble Sort: ordena e conta as operações
def bubble_sort(lista):
    comparacoes = 0
    trocas = 0

    n = len(lista)

    # Percorre a lista
    for i in range(n - 1):

        # Compara os elementos vizinhos
        for j in range(n - 1 - i):

            comparacoes += 1

            # Troca se estiverem fora de ordem
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]
                trocas += 1

    return comparacoes, trocas


# Quick Sort: ordena e conta as operações
def quick_sort(lista):
    comparacoes = 0
    movimentacoes = 0

    # Função que realiza a ordenação
    def ordenar(inicio, fim):
        nonlocal comparacoes, movimentacoes

        # Verifica se ainda há elementos para ordenar
        if inicio < fim:

            # Escolhe o último elemento como pivô
            pivo = lista[fim]
            i = inicio - 1

            # Compara os elementos com o pivô
            for j in range(inicio, fim):

                comparacoes += 1

                # Coloca os menores antes do pivô
                if lista[j] <= pivo:
                    i += 1

                    # Movimenta os elementos
                    if i != j:
                        lista[i], lista[j] = lista[j], lista[i]
                        movimentacoes += 1

            # Coloca o pivô na posição correta
            if i + 1 != fim:
                lista[i + 1], lista[fim] = lista[fim], lista[i + 1]
                movimentacoes += 1

            posicao_pivo = i + 1

            # Ordena a parte esquerda
            ordenar(inicio, posicao_pivo - 1)

            # Ordena a parte direita
            ordenar(posicao_pivo + 1, fim)

    # Inicia o Quick Sort
    ordenar(0, len(lista) - 1)

    return comparacoes, movimentacoes


# Mostra as opções para o usuário
print("=" * 60)
print("       EXPERIMENTO: BUBBLE SORT x QUICK SORT")
print("=" * 60)

print("\nEscolha a quantidade de números:")
print("1 - 10 números")
print("2 - 20 números")
print("3 - 1.000 números")

opcao = input("\nDigite sua opção: ")


# Define a quantidade escolhida
if opcao == "1":
    quantidade = 10

elif opcao == "2":
    quantidade = 20

elif opcao == "3":
    quantidade = 1000

else:
    print("\nOpção inválida!")
    exit()


# Gera os números aleatórios
numeros = [random.randint(0, 1000) for _ in range(quantidade)]


# Cria uma cópia para cada algoritmo
lista_bubble = numeros.copy()
lista_quick = numeros.copy()


# Mostra o array original
print("\n" + "=" * 60)
print(f"ARRAY ORIGINAL ({quantidade} números)")
print("=" * 60)

if quantidade <= 20:
    print(numeros)
else:
    print(numeros[:20], "...")


# Executa o Bubble Sort
bubble_comparacoes, bubble_trocas = bubble_sort(lista_bubble)


# Executa o Quick Sort
quick_comparacoes, quick_movimentacoes = quick_sort(lista_quick)


# Mostra os resultados de cada algoritmo
print("\n" + "=" * 60)
print("RESULTADOS")
print("=" * 60)

print("\nBUBBLE SORT")
print(f"Comparações: {bubble_comparacoes}")
print(f"Trocas:      {bubble_trocas}")

print("\nQUICK SORT")
print(f"Comparações:     {quick_comparacoes}")
print(f"Movimentações:   {quick_movimentacoes}")


# Mostra os arrays ordenados
print("\n" + "=" * 60)
print("ARRAYS ORDENADOS")
print("=" * 60)

if quantidade <= 20:
    print("\nBubble Sort:")
    print(lista_bubble)

    print("\nQuick Sort:")
    print(lista_quick)

else:
    # Mostra apenas os primeiros 20 números
    print("\nBubble Sort (primeiros 20):")
    print(lista_bubble[:20], "...")

    print("\nQuick Sort (primeiros 20):")
    print(lista_quick[:20], "...")


# Mostra a tabela comparativa
print("\n" + "=" * 75)
print("TABELA COMPARATIVA")
print("=" * 75)

print(f"{'Algoritmo':<15}"
      f"{'Comparações':<20}"
      f"{'Trocas/Movimentações':<25}")

print("-" * 75)

print(f"{'Bubble Sort':<15}"
      f"{bubble_comparacoes:<20}"
      f"{bubble_trocas:<25}")

print(f"{'Quick Sort':<15}"
      f"{quick_comparacoes:<20}"
      f"{quick_movimentacoes:<25}")

print("=" * 75)
```

# PARTE 3 – INVESTIGAÇÃO DE BUSCA EM MATRIZES

```python
def busca_sequencial_matriz(matriz, valor_procurado):
    linhas = len(matriz)
    colunas = len(matriz[0])
    comparacoes = 0

    for i in range(linhas):
        for j in range(colunas):
            comparacoes += 1
            if matriz[i][j] == valor_procurado:
                return True, i, j, comparacoes

    return False, -1, -1, comparacoes

# Função auxiliar para criar matrizes de teste
def criar_matriz(linhas, colunas):
    return [[i * colunas + j + 1 for j in range(colunas)] for i in range(linhas)]

# Executando os experimentos
tamanhos = [(2, 2), (10, 10), (100, 100)]

for linhas, colunas in tamanhos:
    matriz = criar_matriz(linhas, colunas)
    total_elementos = linhas * colunas
    print(f"\n--- Matriz {linhas} x {colunas} ({total_elementos} elementos) ---")

    # 1. Valor no início
    valor_inicio = matriz[0][0]
    encontrado, l, c, comp = busca_sequencial_matriz(matriz, valor_inicio)
    print(f"Início [Valor {valor_inicio}]: Encontrado={encontrado}, Linha={l}, Coluna={c}, Comparações={comp}")

    # 2. Valor próximo ao final
    valor_fim = matriz[-1][-1]
    encontrado, l, c, comp = busca_sequencial_matriz(matriz, valor_fim)
    print(f"Fim [Valor {valor_fim}]: Encontrado={encontrado}, Linha={l}, Coluna={c}, Comparações={comp}")

    # 3. Valor inexistente
    valor_inexistente = -999
    encontrado, l, c, comp = busca_sequencial_matriz(matriz, valor_inexistente)
    print(f"Inexistente [Valor {valor_inexistente}]: Encontrado={encontrado}, Linha={l}, Coluna={c}, Comparações={comp}")
```

# PARTE 4 – HANDS ON 1: INVESTIGAÇÃO DO ARRAY

```python
# Cria uma lista (array) vazia para armazenar as 10 temperaturas
temperaturas = []

# 1. Receber as 10 temperaturas
print("Digite 10 temperaturas:")
for i in range(10):
    temp = float(input(f"Temperatura {i + 1}: "))
    temperaturas.append(temp)

# 2. Mostrar todos os elementos armazenados
print("\nTemperaturas armazenadas:", temperaturas)

# 3. Calcular a média
soma = sum(temperaturas)
media = soma / len(temperaturas)
print(f"Média das temperaturas: {media:.2f}")

# 4 e 6. Identificar o maior valor e o seu índice
maior = max(temperaturas)
indice_maior = temperaturas.index(maior)
print(f"Maior valor: {maior} (índice {indice_maior})")

# 5 e 7. Identificar o menor valor e o seu índice
menor = min(temperaturas)
indice_menor = temperaturas.index(menor)
print(f"Menor valor: {menor} (índice {indice_menor})")

# 8. Contar quantos valores estão acima da média
acima_da_media = 0
for t in temperaturas:
    if t > media:
        acima_da_media += 1

print(f"Quantidade de valores acima da média: {acima_da_media}")
```

# PARTE 5 – HANDS ON 2: MATRIZ APLICADA – MONITORAMENTO DE SENSORES

```python
# Inicializando a matriz 5x24 com valores de exemplo
# Linhas = 5 sensores, Colunas = 24 horas
sensores = [
    [22.5, 23.0, 21.5, 20.0, 19.5, 19.0, 20.5, 22.0, 25.0, 27.5, 29.0, 30.5,
     31.0, 30.0, 29.5, 28.0, 27.0, 26.0, 25.0, 24.0, 23.5, 23.0, 22.8, 22.0],
    [20.0, 20.5, 21.0, 19.5, 18.0, 18.5, 19.0, 21.0, 23.0, 25.0, 26.5, 28.0,
     29.0, 28.5, 27.5, 26.0, 25.0, 24.0, 23.0, 22.0, 21.5, 21.0, 20.5, 20.0],
    [25.0, 25.5, 26.0, 24.5, 23.0, 22.5, 23.0, 24.5, 27.0, 29.5, 31.5, 33.0,
     34.0, 33.5, 32.0, 30.5, 29.0, 28.0, 27.0, 26.0, 25.5, 25.0, 24.5, 24.0],
    [18.0, 18.2, 18.5, 17.8, 17.0, 16.5, 17.2, 18.0, 19.5, 21.0, 22.5, 24.0,
     24.5, 24.0, 23.0, 22.0, 21.0, 20.2, 19.8, 19.0, 18.6, 18.4, 18.2, 18.0],
    [21.0, 21.5, 22.0, 20.5, 19.5, 19.0, 20.0, 21.5, 23.5, 25.5, 27.0, 28.5,
     29.5, 29.0, 28.0, 26.5, 25.5, 24.5, 23.5, 22.5, 22.0, 21.8, 21.4, 21.0],
]

NUM_SENSORES = 5
NUM_HORAS = 24

# 1. Média de cada sensor
print("--- 1. Média de cada sensor ---")
for i in range(NUM_SENSORES):
    soma_sensor = 0
    for j in range(NUM_HORAS):
        soma_sensor += sensores[i][j]
    media_sensor = soma_sensor / NUM_HORAS
    print(f"Sensor {i}: {media_sensor:.2f} °C")

# 2, 3 e 4. Maior temperatura, sensor e horário
maior_temp = sensores[0][0]
sensor_maior = 0
horario_maior = 0

soma_total = 0

for i in range(NUM_SENSORES):
    for j in range(NUM_HORAS):
        temp_atual = sensores[i][j]

        # Acumula para a média geral (Item 5)
        soma_total += temp_atual

        # Verifica maior temperatura
        if temp_atual > maior_temp:
            maior_temp = temp_atual
            sensor_maior = i
            horario_maior = j

print("\n--- Estatísticas Globais ---")
print(f"2. Maior temperatura registrada: {maior_temp} °C")
print(f"3. Sensor responsável: Sensor {sensor_maior}")
print(f"4. Horário da ocorrência: {horario_maior}h")

# 5. Média geral
media_geral = soma_total / (NUM_SENSORES * NUM_HORAS)
print(f"5. Média geral: {media_geral:.2f} °C")

# 6. Leituras acima de um limite
limite = float(input("\nDigite o limite de temperatura (°C): "))
contador_acima = 0

for i in range(NUM_SENSORES):
    for j in range(NUM_HORAS):
        if sensores[i][j] > limite:
            contador_acima += 1

print(f"6. Quantidade de leituras acima do limite: {contador_acima}")
```
